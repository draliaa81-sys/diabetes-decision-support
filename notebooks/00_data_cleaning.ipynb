# 00 · Data Cleaning

The Pima Indians Diabetes dataset has some values recorded as **0** that are not possible in real life (for example, a blood pressure or BMI of 0). These zeros really mean "missing".

In this notebook I replace those zeros with the **median** of each column, then save a cleaned copy that the other three notebooks use.

```python
import pandas as pd

df = pd.read_csv("../data/diabetes.csv")
print(df.shape)
df.head()
```

### How many impossible zeros are there?

```python
zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
(df[zero_cols] == 0).sum()
```

### Replace the zeros with the median of each column

```python
df_clean = df.copy()
for col in zero_cols:
    median_value = df[col].median()
    df_clean[col] = df_clean[col].replace(0, median_value)
    print(f"{col}: zeros replaced with {median_value}")

print("\nZeros left:")
print((df_clean[zero_cols] == 0).sum())
```

```python
df_clean.to_csv("../data/diabetes_cleaned.csv", index=False)
print("Saved: data/diabetes_cleaned.csv", df_clean.shape)
```
